# ForeverTAS on Google Colab

Colab's GPU searches for you, for your groups, or for the public pool.

1. **Runtime > Change runtime type > T4 GPU** (or a better one).
2. Put the game's **Packs** folder in your Google Drive, for example `My Drive/ForeverTAS/Packs`
   (the folder from your TrackMania install, with `Stadium.pak` and the others).
3. In ForeverTAS, **Edit > Preferences > Remote**: copy what this machine searches for into `CODES`,
   separated by commas: your own machines' code, groups' invites, `public` for the public pool.
   It searches for one at a time, in that order (your own searches first, the public pool last).
4. **Runtime > Run all.** Colab asks to connect your Drive: allow it.

It stops when Colab ends the session (or when you stop the cell).


```python
CODES = ""  #@param {type:"string"}
PACKS = "/content/drive/MyDrive/ForeverTAS/Packs"  #@param {type:"string"}
NAME = "Colab"  #@param {type:"string"}
# the version your ForeverTAS runs (latest: right after you updated it)
VERSION = "latest"  #@param {type:"string"}
```

```python
from google.colab import drive
drive.mount("/content/drive")
```

```python
# the worker script of the release (the same version as the ForeverTAS that searches)
import shlex
!apt-get -qq install -y zstd libopengl0 libegl1 libxkbcommon0 > /dev/null
release = "latest/download" if VERSION in ("", "latest") else f"download/v{VERSION.lstrip('v')}"
!wget -q -O forevertas_worker.py https://github.com/Skycrafter-dev/ForeverTAS/releases/{release}/forevertas_worker.py
command = shlex.join(["python3", "forevertas_worker.py", "--code", CODES, "--packs", PACKS, "--name", NAME,
                      "--version", VERSION, "--dir", "/content/forevertas-worker"])
!{command}
```
