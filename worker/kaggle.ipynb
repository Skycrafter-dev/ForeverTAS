# ForeverTAS on Kaggle

Kaggle's GPU searches for you, for your groups, or for the public pool.

1. **Settings > Accelerator > GPU**, and **Internet: on**.
2. Add your copy of the game's **Packs** folder as a private dataset (**Add Input**), the folder with
   `Stadium.pak` and the others.
3. In ForeverTAS, **Edit > Preferences > Remote**: copy what this machine searches for into `CODES`,
   separated by commas: your own machines' code, groups' invites, `public` for the public pool.
4. **Run All.**


```python
CODES = ""
NAME = "Kaggle"
VERSION = "latest"   # the version your ForeverTAS runs (latest: right after you updated it)

from pathlib import Path
packs = sorted({p.parent for p in Path("/kaggle/input").rglob("Stadium.pak")})
assert packs, "add your Packs folder as a dataset (Add Input)"
PACKS = str(packs[0])
print("Packs:", PACKS)
```

```python
# the worker script of the release (the same version as the ForeverTAS that searches)
import shlex
!apt-get -qq install -y zstd libopengl0 libegl1 libxkbcommon0 > /dev/null
release = "latest/download" if VERSION in ("", "latest") else f"download/v{VERSION.lstrip('v')}"
!wget -q -O forevertas_worker.py https://github.com/Skycrafter-dev/ForeverTAS/releases/{release}/forevertas_worker.py
command = shlex.join(["python3", "forevertas_worker.py", "--code", CODES, "--packs", PACKS, "--name", NAME,
                      "--version", VERSION, "--dir", "/kaggle/working/forevertas-worker"])
!{command}
```
